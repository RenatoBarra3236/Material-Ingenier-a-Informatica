```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
df=pd.read_csv('Datos.csv')
sns.set_style("whitegrid")
df
```

```python
print("Scatterplot Economy vs. Happiness Score Correlacion ingreso vs. felicidad")
r1=df[["Economy (GDP per Capita)","Happiness Score"]].corr()
sns.lmplot(x="Economy (GDP per Capita)",y="Happiness Score",data=df)
plt.show()
```

*    ¿Que?: Este gráfico de dispersión, con una línea de regresión, tiene como objetivo principal mostrar la correlación entre el ingreso per cápita de un país y su puntaje de felicidad. Revela si existe una relación lineal entre estas dos variables, sugiriendo que a medida que aumenta el PIB per cápita, el puntaje de felicidad también tiende a aumentar.
*    ¿como?: Se usa la marca de la posición para cada punto, donde la posición en el eje x representa el ingreso per cápita y la posición en el eje y representa el puntaje de felicidad. La línea de regresión (con su área de confianza) es un canal de forma que resume la tendencia de los datos, mientras que el sombreado alrededor de ella indica la variabilidad de la relación.
*   ¿porque?: El gráfico de dispersión es la mejor opción para este análisis, ya que es el tipo de gráfico más efectivo para visualizar la relación y correlación entre dos variables cuantitativas. Permite ver la dirección, fuerza y forma de la relación, así como identificar posibles valores atípicos que no se ajusten a la tendencia general.

```python
print("Line Chart Top 10 pa ́ıses vs. Rank Tendencia en puntuaci ́on descendente")
r2= df.sort_values(by='Happiness Rank').head(10)
sns.lineplot(x='Country', y='Happiness Score', data=r2, color="green", marker='o')
plt.xticks(rotation=45, ha='right')
plt.show()
```

*    ¿Que?: Este gráfico de líneas tiene como objetivo principal mostrar la tendencia del puntaje de felicidad para los 10 países mejor clasificados. Al ordenar los países en un ranking, se visualiza claramente la caída gradual en los puntajes de felicidad a medida que se desciende en la lista, lo que resalta la diferencia, aunque pequeña, entre los líderes.
*    ¿cómo?: Se usa la marca de la línea para conectar los puntos de datos, lo que enfatiza la continuidad y dirección de la tendencia. La posición en el eje Y codifica el puntaje de felicidad, mientras que la posición en el eje X representa el país. Esto permite seguir la trayectoria descendente del puntaje de felicidad a lo largo del ranking. El canal de color verde se utiliza para distinguir la línea y los marcadores, mientras que el canal de forma circular se usa para los puntos de datos individuales.
*   ¿por qué?: El gráfico de líneas es la mejor opción para mostrar una tendencia de datos secuenciales como un ranking. A diferencia de un gráfico de barras, la línea continua ayuda a percibir la tasa de cambio de un país a otro y a identificar si la caída es constante o si hay saltos más grandes entre ciertos países.

```python
print("Bar Chart Happiness Score por regi ́on Comparar regiones")
r3=df.groupby('Region')['Happiness Score'].mean().reset_index()
sns.barplot(x='Region', y='Happiness Score', data=r3)
plt.xticks(rotation=45, ha='right')
plt.show()
print("")
```

*   ¿Que?: Este histograma muestra la distribución de los puntajes de felicidad global. Su propósito es visualizar la frecuencia con la que los países caen en rangos específicos de puntajes. Esto revela la forma de la distribución, si es sesgada o si tiene picos, lo que permite entender el comportamiento general del puntaje de felicidad a nivel mundial.
*   ¿cómo?: Se usa la marca de la barra para representar los "bins" o intervalos de puntajes de felicidad. La longitud de la barra (o su altura) codifica la frecuencia de países que se encuentran dentro de ese rango. Las etiquetas en el eje X indican los intervalos de puntaje, mientras que las etiquetas en el eje Y muestran el recuento de países.
*   ¿por qué?: El histograma es la mejor opción para este propósito porque está diseñado para mostrar la distribución de una única variable cuantitativa. A diferencia de un gráfico de barras, las barras del histograma están agrupadas y representan rangos continuos, lo que permite identificar rápidamente la concentración de datos y la forma de la distribución (por ejemplo, si la mayoría de los países tiene un puntaje de felicidad entre 5 y 6).

```python
#javier
print("arallel Coordinates 10 pa´ıses seleccionados Relaciones multivariada")
r11= df[df['Happiness Rank'] <= 10].set_index('Country')
r11= r11.drop('Happiness Rank', axis=1)

fig = px.parallel_coordinates(r11,
                              labels={"Happiness Score": "Puntaje de Felicidad",
                                      "Economy (GDP per Capita)": "Economía (PIB per Cápita)",
                                      "Family": "Apoyo Familiar",
                                      "Health (Life Expectancy)": "Esperanza de Vida",
                                      "Freedom": "Libertad",
                                      "Trust (Government Corruption)": "Confianza",
                                      "Generosity": "Generosidad"},
                              color_continuous_scale=px.colors.diverging.Tealrose,
                              color_continuous_midpoint=5)

fig.update_layout(
    title_text='Relaciones Multivariadas de Factores de Felicidad en Países Seleccionados',
    title_x=0.5,
    legend_title_text='País'
)
fig.show()
```

*   ¿que?: Este gráfico visualiza las relaciones multivariadas entre los factores de felicidad para comparar el perfil de cada país.
Muestra cómo un país tiene valores altos o bajos de forma consistente a lo largo de las variables.
*   ¿como?: Se usa la marca de la línea y el canal de la posición para representar a cada país con una línea que atraviesa ejes que son las variables.
La posición en cada eje indica el valor de la variable, revelando el perfil de datos del país.
*   ¿porque?: Es la mejor opción para comparar perfiles de múltiples países en varias variables a la vez.
Permite identificar fácilmente patrones, correlaciones y diferencias entre los países, algo difícil en gráficos de dos dimensiones.

```python
#javier
print("Pie Chart Factores en pa´ıs top 1 Parte-del-todo")
r12 = df.iloc[0]

r12 = r12.drop(['Country', 'Happiness Score', "Region"])

colores = plt.cm.Paired(range(len(r12)))

plt.figure(figsize=(8, 8))
plt.pie(r12.values,
        labels=r12.index,
        autopct='%1.1f%%',
        startangle=90,
        colors=colores)

plt.title("Contribución de Factores a la Felicidad en Suiza")
plt.ylabel('')
plt.show()
```

*   ¿Que?: Este gráfico de pastel o de tarta visualiza la composición del puntaje de felicidad de un país específico. Su propósito es mostrar la relación de parte-del-todo, es decir, qué porcentaje de la felicidad total es aportado por cada factor, como la economía, la familia y la salud.
*   ¿cómo?: Se utiliza la marca de arco para representar cada factor contribuyente. El canal de ángulo y el canal de área del sector codifican la proporción que cada factor aporta al total. El canal de color diferencia los segmentos para una fácil identificación, y las etiquetas de porcentaje ayudan a cuantificar la contribución de cada parte.
*   ¿por qué?: El gráfico de pastel es la elección ideal para mostrar la composición de un todo. Su diseño es simple y directo, facilitando la comprensión de las proporciones relativas de cada factor en la suma total del puntaje de felicidad, lo cual es difícil de ver con otros tipos de gráficos si el enfoque es un solo país.

```python
#javier
print("polar Area Chart Lo mismo en radial Comparaci´on angular/radial")
r13= df.iloc[0]

# Prepara los datos para el gráfico de área polar
r13= r13.drop(['Country', 'Happiness Score'])

# Crea el gráfico de área polar
fig = px.bar_polar(
    r=r13.values,
    theta=r13.index,
    title=('Contribución de Factores a la Felicidad en suiza')
)

fig.update_traces(
    marker_color=px.colors.sequential.Agsunset,
    text=r13.values,
    hovertemplate="Factor: %{theta}<br>Contribución: %{r:.2f}<extra></extra>"
)

fig.show()
```

*   ¿Que?: Este gráfico de área polar visualiza la contribución de cada factor al puntaje de felicidad total de un país, de manera similar a un gráfico de pastel. El propósito es representar la relación de parte-al-todo de los datos. Sin embargo, este gráfico facilita una mejor comparación visual de las magnitudes de los diferentes factores al utilizar el radio del sector.
*   ¿cómo?: Se usa la marca del arco para representar cada factor, pero a diferencia del gráfico de pastel, la longitud del arco es constante y el canal de codificación es el radio del sector. El ángulo de cada sector es igual y se usa para organizar las categorías. El canal de color se utiliza para diferenciar los sectores, y la longitud radial de cada segmento codifica el valor de cada factor.
*   ¿por qué?: El gráfico de área polar es una excelente alternativa al gráfico de pastel porque, al usar el radio para codificar los valores, el ojo humano puede comparar las diferencias en las magnitudes de forma más precisa. Esto es más efectivo que comparar los ángulos o las áreas de los sectores de un gráfico de pastel, haciendo que las diferencias entre los factores sean más fáciles de percibir.

Esto revela si la distribución es normal, sesgada, o si hay picos que indican concentraciones de países en ciertos niveles de felicidad.
*   ¿cómo?: Se usa la marca de la barra, donde cada barra representa un intervalo de puntajes de felicidad. La longitud de la barra (o su altura) codifica la frecuencia de países que se encuentran dentro de ese rango. Las etiquetas en el eje x indican los intervalos de puntaje, mientras que el eje y muestra el conteo de países.

A diferencia de otros gráficos, agrupa los datos en rangos continuos, lo que permite identificar rápidamente la concentración de países y la forma de la distribución, ofreciendo una visión clara y rápida del panorama global de la felicidad.

```python
#javier
print("Boxplot Happiness Score por regi´on Medianas y outliers")
plt.figure(figsize=(12, 8))
sns.boxplot(data=df, x='Region', y='Happiness Score', palette='viridis')

# Añade el título y etiquetas
plt.title('Distribución de Puntajes de Felicidad por Región')
plt.xlabel('Región')
plt.ylabel('Puntaje de Felicidad')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

*   ¿Que?: Este gráfico de caja y bigotes tiene como propósito principal comparar la distribución de los puntajes de felicidad entre diferentes regiones del mundo. Permite visualizar de forma clara la mediana, la dispersión de los datos y la presencia de valores atípicos (outliers) en cada grupo.
*   ¿cómo?: Se utiliza la marca de la caja y los bigotes, donde la caja representa el rango intercuartílico (IQR), con la línea central marcando la mediana. La posición en el eje Y codifica el puntaje de felicidad, y las líneas (bigotes) muestran la dispersión de los datos, mientras que los puntos individuales marcan los valores atípicos. El canal de color se usa para diferenciar visualmente cada región.
*   ¿por qué?: Un gráfico de caja y bigotes es la mejor opción para este propósito porque resume la distribución de los datos de manera compacta y visualmente potente. Permite una comparación rápida y eficaz de métricas clave como la mediana, la variabilidad y los outliers entre múltiples grupos (regiones), lo cual es más difícil de lograr con histogramas o gráficos de barras.
