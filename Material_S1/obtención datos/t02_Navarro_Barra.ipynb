```python
def contar(texto):
    quitar=",;:.\n!\"'" #definimos lo que vamos a quitar

    for caracter in quitar: #remplazamos
        texto= texto.replace(caracter,"")

    texto=texto.lower() #transformar todo a minusculas
    texto_sin= texto.split(" ") #separamos las palabras

    d=dict() #contamos las palabras
    for i in texto_sin:
        if i in d:
            d[i]+=1
        else:
            d[i]=1

    serie_calculo= pd.Series(d) #creamos la serie_calculo

    d_invertido={} #invertimos la matriz
    for palabra, frecuencia in d.items():
        if frecuencia in d_invertido:
            d_invertido[frecuencia].append(palabra)
        else:
            d_invertido[frecuencia]=[palabra]

    datos= [(frecuencia,palabra) for frecuencia, palabras in d_invertido.items() for palabra in palabras] #extendemos para que se vean todas
    
    frecuencias, palabras=zip(*datos) #crear serie con frecuencia como indice
    serie_mostrar=pd.Series(palabras,index=frecuencias) #creamos la serie con frecuencia, indice

    maximo= serie_mostrar[serie_mostrar.index==serie_mostrar.index.max()] #maximo
    minimo= serie_mostrar[serie_mostrar.index==serie_mostrar.index.min()] #minimo

    p_unicas= (pd.Series(d)==1).sum()
    p_repetidas= (pd.Series(d) > 1).sum()

    mediana= pd.Series(list(d.values())).median()
    numero= None
    for palabra, frecuencia in d.items():
        if frecuencia == mediana:
            numero = palabra
            break

    return serie_mostrar, maximo, minimo, p_unicas, p_repetidas, numero

texto=("El sol brillaba en el cielo, mientras los niños jugaban en el parque. Algunos pájaros cantaban, creando una melodía alegre. Carlos observaba desde un banco, recordando sus días de infancia. La pelota de un niño rodó hacia él, y con una sonrisa, la devolvió. La alegría de los niños le hizo sentir nostalgia. Mientras tanto, el viento movía suavemente las hojas de los árboles.")

serie_mostrar,maximo,minimo,p_unicas, p_repetidas, numero=contar(texto)


print(f"esta es nuestra serie: \n{serie_mostrar}")
print()
print(f"esta es nuestro maximo: \n{maximo}")
print()
print(f"esta es nuestro minimo: \n{minimo}")
print()
print(f"esta es nuestra palabra unica: {p_unicas}")
print()
print(f"esta son nuestras palabras repetidas: {p_repetidas}")
print()
print(f"esta es nuestro mediana: {numero}")


```

```python
import random

#n= random.randint(15,26)

def usuarios(n):
    d = {}
    for i in range(n):
        nombre = input("Ingrese su nombre: ")
        edad = int(input("Ingrese su edad: "))
        genero = input("Ingrese F o M: ")
        estatura = float(input("Estatura en cm: "))
        peso = float(input("Peso en kg: "))

        d[nombre] = (edad, genero, estatura, peso)

    # Creación del DataFrame
    df = pd.DataFrame(d, index=["edad", "genero", "estatura", "peso"]).T

    # Calcular IMC
    df["imc"] = round(df["peso"] / ((df["estatura"] / 100) ** 2), 1)

    # Clasificar la salud según el IMC
    for i in range(len(df)):
        if df["imc"].iloc[i] < 18.5:
            df.at[df.index[i], "salud"] = "Bajo peso"
        elif 18.5 <= df["imc"].iloc[i] <= 24.9:
            df.at[df.index[i], "salud"] = "Normal"
        elif 25 <= df["imc"].iloc[i] <= 29.9:
            df.at[df.index[i], "salud"] = "Sobrepeso"
        else:
            df.at[df.index[i], "salud"] = "Obesidad"

    # DataFrames separados para cada categoría de salud
    df_bp = df[df["salud"] == "Bajo peso"]
    df_nm = df[df["salud"] == "Normal"]
    df_sp = df[df["salud"] == "Sobrepeso"]
    df_ob = df[df["salud"] == "Obesidad"]

    # Función para calcular estadísticas por grupo
    def calcular_estadisticas(df_grupo):
        if len(df_grupo) > 0:
            estadisticas = {
                "promedio_estatura": round(df_grupo["estatura"].mean(), 2),
                "min_estatura": df_grupo["estatura"].min(),
                "max_estatura": df_grupo["estatura"].max(),
                "moda_estatura": df_grupo["estatura"].mode().values[0],
                "std_estatura": round(df_grupo["estatura"].std(), 2),

                "promedio_peso": round(df_grupo["peso"].mean(), 2),
                "min_peso": df_grupo["peso"].min(),
                "max_peso": df_grupo["peso"].max(),
                "moda_peso": df_grupo["peso"].mode().values[0],
                "std_peso": round(df_grupo["peso"].std(), 2),

                "promedio_imc": round(df_grupo["imc"].mean(), 2),
                "min_imc": df_grupo["imc"].min(),
                "max_imc": df_grupo["imc"].max(),
                "moda_imc": df_grupo["imc"].mode().values[0],
                "std_imc": round(df_grupo["imc"].std(), 2)
            }
            return estadisticas
        else:
            return "No hay datos en este grupo."

    print("\nDataFrame original:")
    print(df)

    print("\nDataFrame de Bajo peso:")
    print(df_bp)
    print("Estadísticas de Bajo peso:")
    print(calcular_estadisticas(df_bp))
    
    print("\nDataFrame de Normal:")
    print(df_nm)
    print("Estadísticas de Normal:")
    print(calcular_estadisticas(df_nm))
    
    print("\nDataFrame de Sobrepeso:")
    print(df_sp)
    print("Estadísticas de Sobrepeso:")
    print(calcular_estadisticas(df_sp))
    
    print("\nDataFrame de Obesidad:")
    print(df_ob)
    print("Estadísticas de Obesidad:")
    print(calcular_estadisticas(df_ob))

print(usuarios(2))
```
