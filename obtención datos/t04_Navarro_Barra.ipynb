```python
def mercados(market_code):
   global df_estadistico
   global caracteristicas #Esto esta para poder usarlos en el resto del codigo
   global df_mercado
   global mercado 

   mercado = market_code   
   print(f"Análisis de características musicales para {market_code}:")
   df_mercado = df[df['market_code'] == market_code]
   caracteristicas = ['danceability', 'energy', 'valence', 'speechiness', 'tempo','duration_ms']
   df_estadistico= df_mercado[caracteristicas].describe().T

   print(f'Top 3 canciones populares para {market_code}')
   display(df_mercado[['track_name','artist_name','popularity','duration_ms']].sort_values('popularity', ascending=False).head(3))

   return df_estadistico
   
display(mercados("CL"))
```

```python
def baile(df,fila,columna, df2): #el df2 esta dado que se necesita un df donde el danceability sea una columna
   promedio=df.loc[fila, columna]
   Q1= df2[fila].quantile(0.25)
   Q2= df2[fila].quantile(0.50)
   if promedio< Q1:
      print(f"El nivel de danceability es {promedio} por lo tanto es bajo \n")
   elif Q1<=promedio< Q2:
      print(f"El nivel de danceability es {promedio} medio \n")
   else:
      print(f"El nivel de danceability es {promedio} alto \n")
baile(df_estadistico, "danceability", "mean", df_mercado)

def palabras(df, fila, columna, min, max):
   promedio=df.loc[fila, columna]
   minimo= df.loc[fila, min]
   maximo= df.loc[fila, max]
   rango= maximo - minimo
   if promedio < rango:
      print(f"En speechiness el promedio: {promedio} esta bajo el nivel del rango: {rango} el valor minimo es {minimo} y el valor maximo es {maximo} \n")
   elif promedio == rango:
      print(f"En speechiness el promedio: {promedio} es igual al rango: {rango} el valor minimo es {minimo} y el valor maximo es {maximo} \n")
   else:
      print(f"En speechiness el promedio: {promedio} esta sobre el nivel del rango: {rango} el valor minimo es {minimo} y el valor maximo es {maximo} \n")
   
palabras(df_estadistico, "speechiness","mean", "min", "max")

def tiempo(df,columna):
   columna_duracion= df[columna]
   columna_duracion=round(columna_duracion/60000,2)
   promedio= columna_duracion.mean()
   return promedio

tiempo(df_mercado,"duration_ms" )
```

```python
# b) Visualización básica usando pandas
# Solo una visualización simple por mercado
caracteristicas_sin_duracion = [c for c in caracteristicas if c not in ['duration_ms','tempo']]

df_mercado[caracteristicas_sin_duracion].boxplot(vert=False)  
plt.title(f'Características musicales en {mercado}')
plt.xlabel("Valor")
plt.show()
```

# Codigo para comparar medias #

```python
#Codigo que compara la moda en los 3 mercados

mercados_comparar = ["CL", "ES", "GB"]
columnas_comparar = ["tempo","artist_genres"]

def comparar_moda(df, mercados, columnas):
    modas = {}
    for mercado in mercados:
        df_mercado = df[df['market_code'] == mercado]  # Filtrar los datos por mercado
        modas_mercado = {}
        
        for columna in columnas:
            if columna == "artist_genres":
                generos = df_mercado[columna].str.split(",", expand=True).stack()  # Calcular la moda para cada columna
                moda= generos.mode()
                modas_mercado[columna] = moda.values[0]  #el values[0] es para tomar el primero
            else:
                moda = df_mercado[columna].mode()
                modas_mercado[columna] = moda.values[0]
        
        modas[mercado] = modas_mercado
    
    return pd.DataFrame(modas)
display(comparar_moda(df, mercados_comparar, columnas_comparar))
```

# Item 4 #

Comparando con las tendencias globales, observamos que:
1. {Mercado 1} se destaca globalmente en {característica},
   superando la media global por {porcentaje}%
2. {Mercado 2} muestra una desviación significativa en {característica},
   que podría explicarse por {razón}
3. Los tres mercados principales comparten {característica común}
   que los diferencia del promedio global en {medida}

```python
dfs = []
global_means = df[caracteristicas].mean()
for mercado in ['CL', 'ES', 'GB']:
    mercado_means = df[df['market_code'] == mercado][caracteristicas].mean()
    diff_percent = ((mercado_means - global_means) / global_means * 100).round(2)
    print(f"\nDiferencia porcentual con la media global para {mercado}:")
    dfs.append(diff_percent)
    print(diff_percent)
```

```python
# 1)
caracteristica = dfs[0].abs().idxmax()
porc = dfs[0].abs()[caracteristica]
print(f'1.- El mercado de Chile se destaca globalmente en {caracteristica}, superando la media global por {porc}%.\n')

#2)
carac2 = dfs[1].abs().idxmax()

print(f'2.- El mercado de España muestra una desviación significativa en {carac2}, podría explicarse por una mayor preferencia cultural por canciones rítmicas y fáciles de bailar.\n')

# 3)
comun = (dfs[0] + dfs[1] + dfs[2]).abs().idxmax()
med = (dfs[0] + dfs[1] + dfs[2]).abs().max()
print(f'3.- Los tres mercados principales comparten "{comun}" que los diferencia del promedio global en {-med}% de diferencia porcentual total.')
```
